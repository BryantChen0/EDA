# Exploratory Data Analysis of Food Nutrition Data

## Introduction

This project explores a food nutrition dataset containing information about different foods and their nutritional characteristics.

The goal of this exploratory data analysis is to understand the structure and distribution of the data, identify patterns and relationships between variables, and detect potential outliers.

Import Request lib

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Dataset overview

```python
df.describe()
```

From the data overview, the numerical variables include `id`, `calories`, `protein`, `carbs`, `fat`, `fiber`, `sugar`, `saturated_fat`, and `salt`, while `name`, `category`, and `brand` are categorical variables. Although `id` is stored as a numerical value, it is an identifier and does not provide meaningful information for numerical analysis. Therefore, I excluded `id` from the numerical analysis. The `brand` variable also contains many missing values, so it will not be a major focus of the analysis.

The main numerical variables of interest are `calories`, `protein`, `carbs`, `fat`, `fiber`, `sugar`, `saturated_fat`, and `salt`. The counts for `calories`, `protein`, `carbs`, `fat`, and `fiber` are higher than those for `sugar`, `saturated_fat`, and `salt`. This indicates that the latter variables contain more missing values in the dataset, with `sugar` having a particularly large amount of missing data.

```python
df['category'].value_counts().plot(kind='barh')

plt.xlabel('Count')
plt.ylabel('Category')
plt.title('Category Distribution')

plt.show()
```

```python
df['source'].value_counts().plot(kind='barh')

plt.xlabel('Count')
plt.ylabel('Source')
plt.title('Source Distribution')

plt.show()
```

The graph of `carbs`, `fat` and `calories` may indicate skewed distributions or the presence of extreme values. These potential outliers will be examined further using box plots. Based on the standard deviations, `salt` has the smallest variation among the numerical variables, indicating that its values are relatively close to each other compared with the other variables. For the Category Distribution graph, the Liha ja linnuliha category have a lot more than other category.

Outlier analysis through the boxplt and IQR

```python
df[['calories']].boxplot()
```

```python
df[['protein', 'carbs', 'fat', 'sugar', 'saturated_fat']].boxplot()
```

```python
df[['fiber', 'salt']].boxplot()
```

From the previous three boxplots, all the outliers are higher than the other data points. As shown in the previous analysis, the boxplot of carbs has more outliers than the other numerical variables. The boxplots of sugar, fat, saturated_fat, fiber, and salt also show many outliers. One difference among these variables is that salt and carbs have more closely grouped outliers, while the outliers of the other numerical variables are more widely distributed.
