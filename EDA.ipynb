Outlier analysis through the boxplt and IQR

```python
df[['calories']].boxplot()
```

```python
df[['protein', 'carbs', 'fat', 'sugar', 'saturated_fat']].boxplot()
```

```python
df[['fiber', 'salt']].boxplot()
```

From the previous three boxplots, all the outliers are higher than the other data points. As shown in the previous analysis, the boxplot of carbs has more outliers than the other numerical variables. The boxplots of sugar, fat, saturated_fat, fiber, and salt also show many outliers. One difference among these variables is that salt and carbs have more closely grouped outliers, while the outliers of the other numerical variables are more widely distributed.

Pair-wise scatter plot matrix

```python
numeric_cols = [
    'calories',
    'protein',
    'carbs',
    'fat',
    'fiber',
    'sugar',
    'saturated_fat',
    'salt'
]

sns.pairplot(df[numeric_cols])
plt.show()
```

According to the previous pair-wise scatter plot matrix, there are few relationship between most of them. Calories have a strong positive relationship with fat and a moderate positive relationship with carbs and protein. Carbs and sugar also show a clear positive association, while total fat and saturated fat are strongly positively related. Fiber and salt show weaker relationships with most of the other nutritional variables. Overall, the scatter plot matrix suggests that some nutritional variables are strongly related, while others have relatively weak associations.

```python
sns.boxplot(
    data=df,
    x='category',
    y='calories',
    hue='category',
    legend=False
)

plt.xlabel('Category')
plt.ylabel('Calories')
plt.title('Distribution of Calories Across Categories')
plt.xticks(rotation=90)
plt.show()
```

According to the box plot above, the `Rasvad ja õlid` category has a higher median calorie value than the other categories and does not appear to have any outliers. This category also has the largest variation in calorie values. The `Liha ja linnuliha` category has the most outliers, while its variation is only slightly smaller than that of the `Rasvad ja õlid` category.
